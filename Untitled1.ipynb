```python
df_agg_bookings["occ_pct"] = df_agg_bookings["successful_bookings"]/df_agg_bookings["capacity"]#we are creating a new column of booking percentage
df_agg_bookings.head()
```

```python
df_agg_bookings["occ_pct"] = df_agg_bookings["occ_pct"].apply(lambda x: round(x*100 , 2))
df_agg_bookings.head(4)
df_agg_bookings.head(10)
df_agg_bookings.head(100)
```

```python
df_agg_bookings.groupby("room_category")["occ_pct"].mean().round(2)
```

```python
df_rooms
```

```python
df = pd.merge(df_agg_bookings,df_rooms,left_on = "room_category",right_on = "room_id")
df.tail(4)  #joining two different dataframe
```

```python
df.groupby("room_class")["occ_pct"].mean().round(2)
```

```python
df.drop("room_id", axis=1, inplace=True, errors="ignore")
df.head(4)  #deleted the column of room_id
```
